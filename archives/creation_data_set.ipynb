But du script : prendre l'ancien fichier csv et en créer un nouveau en regroupant des sentiments 

```python
import pandas as pd
```

```python
dataset = 'test-english'
# dataset = 'training-english'
# dataset = 'validation-english'
```

```python
file_path = f"datasets/valueeval24/{dataset}/labels.tsv"
```

### 19 valeurs sans attained/ constrained

```python
output_file = f"datasets/valueeval24/{dataset}/values_labels.tsv"
df = pd.read_csv(file_path, sep='\t')

values = [ "Self-direction: thought", "Self-direction: action", "Stimulation",  "Hedonism", "Achievement", "Power: dominance", "Power: resources", "Face", "Security: personal", "Security: societal", "Tradition", "Conformity: rules", "Conformity: interpersonal", "Humility", "Benevolence: caring", "Benevolence: dependability", "Universalism: concern", "Universalism: nature", "Universalism: tolerance" ]

groupes = {
    value : [f"{value} attained", f"{value} constrained"] for value in values
} 

output = pd.DataFrame()
output["Text-ID"] = df["Text-ID"]	
output["Sentence-ID"] = df["Sentence-ID"]
for nom_groupe, colonnes in groupes.items():
    output[nom_groupe] = df[colonnes].any(axis=1).astype(int) 
    
output["No Value"] = (output[list(groupes.keys())].sum(axis=1) == 0).astype(int)

output.to_csv(output_file, sep='\t', index=False)

print(f"Fichier {output_file} créé avec succès !")
```
