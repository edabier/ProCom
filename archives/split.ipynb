```python
import pandas as pd 

#input : csv avec les artciles copier coller
#output : comme sentences.tsv mais en csv 
```

```python

data= pd.read_csv("input.csv")
data.dropna(inplace=True)

output = pd.DataFrame(columns=["Text-ID","Sentence-ID","Text"])

for index, row in data.iterrows():
    print(row["Text"])
    row["Text"] = row["Text"].split(".")
    phrases = row["Text"]
    for i in range(len(phrases)-1) :
        new_row = pd.DataFrame({"Text-ID":[row["Text-ID"]],"Sentence-ID":[i],"Text": [phrases[i]]})
        output = pd.concat([output,new_row],ignore_index=True)

output.to_csv("output.tsv",sep='\t')
    
# df display
```
